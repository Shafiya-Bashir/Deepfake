# 02. Data Quality & Preprocessing Audit
Auditing facial detection confidence, bounding box margins, Laplacian blur scores, and frame rates.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/processed/clean_metadata.csv')
print(f"Face detection success rate: {df['face_detected'].mean() * 100:.1f}%")
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df['laplacian_blur_score'], bins=30, kde=True, color='#1f77b4')
plt.axvline(50, color='red', linestyle='--', label='Blur Filter Threshold (50)')
plt.title('Distribution of Laplacian Variance (Sharpness) Scores')
plt.xlabel('Sharpness Variance')
plt.legend()
plt.show()
```
