# 03. Frequency Domain & Feature Analysis
Analyzing 2D-DCT frequency spectra and high-frequency residual energy across authentic and generated faces.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fftpack import dct

# Synthetic demonstration of 2D-DCT spectral energy
real_spectrum = np.random.exponential(scale=1.0, size=(64, 64))
fake_spectrum = real_spectrum.copy()
fake_spectrum[32:, 32:] += np.random.normal(loc=2.5, scale=0.5, size=(32, 32)) # Generator artifact spike

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(np.log1p(real_spectrum), cmap='inferno')
axes[0].set_title('Authentic Image 2D-DCT Spectrum')
axes[1].imshow(np.log1p(fake_spectrum), cmap='inferno')
axes[1].set_title('Synthetic Image (Notice High-Freq Checkerboard Spikes)')
plt.show()
```
