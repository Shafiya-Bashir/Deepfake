# 01. Data Exploration & Dataset Distributions
Exploratory data analysis comparing real vs synthetic facial media across FaceForensics++, Celeb-DF, and DFDC.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/processed/clean_metadata.csv')
print(f'Total video samples loaded: {len(df)}')
df.head()
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='dataset', hue='label_name', palette={'REAL': '#2ca02c', 'FAKE': '#d62728'})
plt.title('Real vs. Synthetic Class Distribution by Dataset')
plt.ylabel('Video Count')
plt.show()
```
